```python
#Build a simple sentiment analyzer and test on 5 reviews.
positive_words = [
    "good", "great", "excellent", "amazing",
    "awesome", "fantastic", "love", "best"
]

negative_words = [
    "bad", "worst", "boring", "poor",
    "terrible", "hate", "awful", "disappointing"
]

reviews = [
    "This movie was amazing and awesome",
    "The film was boring and terrible",
    "I love this movie, it is fantastic",
    "The movie was bad and disappointing",
    "It was a great movie with excellent acting"
]

for i, review in enumerate(reviews, start=1):

    review = review.lower()

    pos = 0
    neg = 0

    words = review.split()

    for word in words:
        if word in positive_words:
            pos += 1

        if word in negative_words:
            neg += 1

    if pos > neg:
        sentiment = "Positive"
    elif neg > pos:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Review", i)
    print("Text:", review)
    print("Sentiment:", sentiment)
    print()
```

```python
#Build a simple sentiment analyzer and test on 5 reviews.
import re
try:
    from textblob import TextBlob
except ModuleNotFoundError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "textblob"])
    from textblob import TextBlob
    try:
        from textblob import download_corpora
        download_corpora.download_all()
    except Exception:
        pass
def analyze_sentiment(review):
    # Clean the review text
    review = re.sub(r'[^\w\s]', '', review)  # Remove punctuation
    review = re.sub(r'\d+', '', review)  # Remove numbers
    review = review.lower()  # Convert to lowercase
    
    # Analyze sentiment using TextBlob
    analysis = TextBlob(review)
    
    if analysis.sentiment.polarity > 0:
        return "Positive"
    elif analysis.sentiment.polarity < 0:
        return "Negative"
    else:
        return "Neutral"
# Sample reviews
reviews = [
    "I love this product! It works great and exceeded my expectations.",
    "This is the worst purchase I've ever made. Completely useless.",
    "It's okay, not bad but not great either.",
    "Amazing quality and fantastic customer service!",
    "Terrible experience, the item arrived broken and support was unhelpful."
]
# Analyze each review and print the result
for i, review in enumerate(reviews, 1):
    sentiment = analyze_sentiment(review)
    print(f"Review {i}: {sentiment}")

    
```
