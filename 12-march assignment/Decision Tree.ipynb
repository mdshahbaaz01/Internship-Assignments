```python
sunny = input("Is it sunny? (yes/no): ").lower()

if sunny == "no":
    print("Play Outside")
else:
    temp = int(input("Enter temperature: "))

    if temp > 35:
        print("Don't Play Outside")
    else:
        weekend = input("Is it weekend? (yes/no): ").lower()

        if weekend == "yes":
            print("Play Outside")
        else:
            print("Don't Play Outside")
```
