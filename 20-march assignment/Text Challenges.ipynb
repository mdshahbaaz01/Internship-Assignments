```python
#Collect 20 messy sentences and identify slang, emojis, typos; explain preprocessing needed.
messy_text = [
    "Hiiiii brooo 😄",
    "Gud mrng frnd",
    "I luv this movie ❤️",
    "Wat r u doing?",
    "OMG!!! This is awesome 😍",
    "Plz send d notes",
    "Thnks alottt 😊",
    "C u tomorrow",
    "LOL that's funny 😂",
    "Hpy bdayyy 🎂",
    "I canttt understand dis",
    "Gr8 work buddy 👍",
    "Whr r u nw?",
    "This phn is amazng",
    "Tysm for ur help ❤️",
    "M sooo exciteddd 😁",
    "Btw, meet me @ 5pm",
    "U r the besttt",
    "Congratzzz 🎉🎉",
    "Gm everyone 😊"
]

slang_words = [
    "luv", "plz", "gr8", "lol", "tysm",
    "btw", "gm", "wat", "ur", "u"
]

emoji_list = [
    "😄", "❤️", "😍", "😊", "😂",
    "🎂", "👍", "😁", "🎉"
]

print("Text Analysis\n")

for i, txt in enumerate(messy_text, start=1):

    issues = []

    # check slang
    for word in slang_words:
        if word in txt.lower():
            issues.append("Slang")
            break

    # check emojis
    for emoji in emoji_list:
        if emoji in txt:
            issues.append("Emoji")
            break

    # simple typo check
    if any(ch * 3 in txt.lower() for ch in "abcdefghijklmnopqrstuvwxyz"):
        issues.append("Repeated Letters / Typo")

    print(i, ".", txt)
    print("Issues Found:", ", ".join(issues))
    print()
```
