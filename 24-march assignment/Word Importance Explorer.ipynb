```python
#Use TF-IDF on 5 documents and identify top keywords with explanation.

from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

docs = [
    "Artificial intelligence is changing the world",
    "Machine learning is a part of artificial intelligence",
    "Deep learning improves AI applications",
    "Artificial intelligence helps in automation",
    "Machine learning and deep learning are popular"
]

tfidf = TfidfVectorizer()

mat = tfidf.fit_transform(docs)

words = tfidf.get_feature_names_out()

df = pd.DataFrame(mat.toarray(), columns=words)

print("TF-IDF Matrix:\n")
print(df)

print("\nTop Keywords in Each Document:\n")

for i in range(len(docs)):
    row = df.iloc[i]

    top_words = row.sort_values(ascending=False).head(3)

    print("Document", i + 1)
    print("Text:", docs[i])
    print("Keywords:", list(top_words.index))
    print()
```
