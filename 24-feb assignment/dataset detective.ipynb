```python
#Load a dataset, display top rows, find highest value column, count missing values, share 5 insights.
import pandas as pd

# loading dataset (replace file name if needed)
df = pd.read_csv("student_data.csv")

print("Top 5 rows:")
print(df.head())

print("\nColumn info:")
print(df.info())

# highest value column (numeric only)
num_df = df.select_dtypes(include=['int64', 'float64'])

max_cols = num_df.max()

print("\nMax values in each numeric column:")
print(max_cols)

# column with highest single value
highest_col = max_cols.idxmax()
print("\nColumn with highest value:", highest_col)

# missing values count
print("\nMissing values per column:")
print(df.isnull().sum())

# quick insights (basic)
print("\n--- Insights ---")

print("1. Dataset has", df.shape[0], "rows and", df.shape[1], "columns")

print("2. Numeric columns:", list(num_df.columns))

print("3. Column with highest value overall:", highest_col)

print("4. Total missing values:", df.isnull().sum().sum())

print("5. Most filled column:", df.isnull().sum().idxmin())
```
