```python
#create a strong code for password authentication in python
password = input("Enter your password: ")
has_upper = False
has_lower = False
has_digit = False

for ch in password:
    if "A" <= ch <= "Z":
        has_upper = True
    if "a" <= ch <= "z":
        has_lower = True
    if "0" <= ch <= "9":
        has_digit = True

if len(password) >= 9 and has_upper and has_lower and has_digit:
    print("Password is strong.")

else:
    print("weak password, try again.") 
```
