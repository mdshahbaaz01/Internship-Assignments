```python
#create shopping cart using each and remove one item and print total items

shopping_list = input("Enter the item names separated by commas: ").split(",")
shopping_items = list(shopping_list)
print(shopping_items)

remove_item = input("Enter the item you want to remove: ")
shopping_items.remove(remove_item)
print("You removed this: ",remove_item)
print(shopping_items)

item_list = set(shopping_items)

full_list = item_list 
for item_list in full_list:
    print("This is the item list: ",item_list)
```
