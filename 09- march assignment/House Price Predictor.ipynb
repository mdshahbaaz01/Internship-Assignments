```python
#Train a Linear Regression model, predict prices, and test with new input.
from sklearn.linear_model import LinearRegression
import pandas as pd

# sample dataset
data = {
    "Area": [500, 700, 900, 1100, 1300, 1500],
    "Price": [10, 14, 18, 22, 26, 30]
}

df = pd.DataFrame(data)

# feature and label
X = df[["Area"]]
y = df["Price"]

# train model
model = LinearRegression()
model.fit(X, y)

# prediction on existing data
pred = model.predict(X)

print("Predicted Prices:")
for i in range(len(X)):
    print("Area:", X.iloc[i, 0], "-> Price:", round(pred[i], 2))

# test with new input
new_area = [[1200]]
new_price = model.predict(new_area)

print("\nNew House Prediction")
print("Area:", new_area[0][0])
print("Predicted Price:", round(new_price[0], 2))
```

```python
#taking data from user
from sklearn.linear_model import LinearRegression
import pandas as pd

area_list = []
price_list = []

n = int(input("How many house records do you want to enter? "))

for i in range(n):
    print("\nRecord", i + 1)

    area = int(input("Enter Area (sq ft): "))
    price = int(input("Enter Price (in lakhs): "))

    area_list.append(area)
    price_list.append(price)

# create dataset
df = pd.DataFrame({
    "Area": area_list,
    "Price": price_list
})

print("\nDataset:")
print(df)

# feature and label
X = df[["Area"]]
y = df["Price"]

# train model
model = LinearRegression()
model.fit(X, y)

# test with new input
new_area = int(input("\nEnter area to predict price: "))

pred_price = model.predict([[new_area]])

print("Predicted Price:", round(pred_price[0], 2), "Lakhs")

# print(df.head())  # debug
```
