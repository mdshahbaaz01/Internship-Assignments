```python
#Write code to remove punctuation, lowercase text, remove stopwords and test it.

import string

text = input("Enter a sentence: ")

# convert to lowercase
text = text.lower()

# remove punctuation
for ch in string.punctuation:
    text = text.replace(ch, "")

# stopwords list
stop_words = [
    "is", "am", "are", "the", "a", "an",
    "and", "or", "of", "to", "in", "on",
    "for", "with", "this", "that"
]

words = text.split()

clean_words = []

for word in words:
    if word not in stop_words:
        clean_words.append(word)

clean_text = " ".join(clean_words)

print("\nCleaned Text:")
print(clean_text)
```
