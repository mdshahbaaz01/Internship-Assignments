```python
from sklearn.neighbors import NearestNeighbors
import pandas as pd

data = {
    "Action": [5, 5, 1],
    "Comedy": [4, 5, 2],
    "Romance": [1, 2, 5]
}

users = ["A", "B", "C"]

df = pd.DataFrame(data)

knn = NearestNeighbors(n_neighbors=2)
knn.fit(df)

dist, ind = knn.kneighbors([df.iloc[0]])

print("User A Similar Users:")

for i in ind[0]:
    print(users[i])
```
