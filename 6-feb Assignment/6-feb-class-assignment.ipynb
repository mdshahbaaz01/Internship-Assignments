```python
#Create a function that squares a number
def square(num):
    return num*num
num = 7
result = square(num)
print("The square of the number is: ", result)
```

```python
#Create a function that squares a number (taking input from the user)
def square(num):
    return num**2
num = int(input("Enter a number: "))
result = square(num)
print("The square of the number is: ", result)
```
