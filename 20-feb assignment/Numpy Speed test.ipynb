```python
#Compare Python lists vs NumPy arrays with 1M numbers, measure execution time, write 3 observations
import time
import numpy as np

# 1 million numbers
n = 1000000

# python list
list1 = list(range(n))

# numpy array
arr = np.arange(n)

# ---- list operation timing ----
start = time.time()
list_sum = 0
for i in list1:
    list_sum += i
end = time.time()

list_time = end - start

# ---- numpy operation timing ----
start = time.time()
arr_sum = np.sum(arr)
end = time.time()

numpy_time = end - start

print("List Time:", list_time)
print("NumPy Time:", numpy_time)

print("List Sum:", list_sum)
print("NumPy Sum:", arr_sum)
```
