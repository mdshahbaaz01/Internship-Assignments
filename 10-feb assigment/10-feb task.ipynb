```python
#1. print the number is even or odd
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Number is Even")
else:
    print("Number is Odd")
```

```python
#Sum until user stops (enter 0 to stop)
total = 0

while True:
    num = int(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    total += num

    # print("current sum:", total)

print("Final Sum =", total)
```
