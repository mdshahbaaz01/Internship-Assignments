```python

import pandas as pd

# Read the CSV file
df = pd.read_csv('student_records.csv') 
print("**Original Data**")
print(df, "\n")

# Drop rows where 'name' or 'dept' is missing
df.dropna(subset=['name', 'dept'], inplace=True) 

# Fill missing numerical values with the mean of their respective columns
df['maths'] = df['maths'].fillna(df['maths'].mean())
df['science'] = df['science'].fillna(df['science'].mean())
df['english'] = df['english'].fillna(df['english'].mean())

# Add a new column 'total'
df['total'] = df['maths'] + df['science'] + df['english'] 

# Add new column 'avg'
df['avg'] = df['total'] / 3 
print("**Data with Total and Avg**")
print(df.head(), "\n")

# Find top 3 students based on 'total'
top_3_students = df.nlargest(3, 'total') 
print("**Top 3 Students**")
print(top_3_students[['name', 'dept', 'total']], "\n")

# Department-wise average marks
dept_wise_avg = df.groupby('dept')['avg'].mean().reset_index()
dept_wise_avg.rename(columns={'avg': 'dept_overall_avg'}, inplace=True)

print("**Department wise Average Marks**")
print(dept_wise_avg, "\n")
```
