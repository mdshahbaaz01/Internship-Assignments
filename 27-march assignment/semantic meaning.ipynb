```python
#Find 5 word pairs and explain semantic similarity
word_pairs = {
    "Car - Automobile": "Both are vehicles",
    "Happy - Joyful": "Both express positive emotions",
    "Student - Learner": "Both relate to learning",
    "Big - Large": "Both describe size",
    "Teacher - Instructor": "Both refer to teaching"
}

print("Semantic Similarity Examples\n")

for pair, meaning in word_pairs.items():
    print(pair)
    print("Explanation:", meaning)
    print()
```
