```python
#1. create numpy array with the following temp in degree celsius [28,32,40,37,26,38]
#requirements: convert the list into numpy array, print maximum and minimum temperature,and calculate average temp and display last 3 days temp 
import numpy as np

temps = [28, 32, 40, 37, 26, 38]

arr = np.array(temps)

print("Temperature Array:", arr)

print("Maximum Temperature:", arr.max())
print("Minimum Temperature:", arr.min())

avg_temp = arr.mean()
print("Average Temperature:", avg_temp)

print("Last 3 Days Temperature:", arr[-3:])
```
