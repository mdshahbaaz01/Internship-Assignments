```python
msg = input("Enter a message: ")

spam_words = [
    "free",
    "winner",
    "prize",
    "offer",
    "click",
    "cash",
    "gift",
    "congratulations"
]

msg = msg.lower()

is_spam = False

for word in spam_words:
    if word in msg:
        is_spam = True
        break

if is_spam:
    print("Spam Message")
else:
    print("Not Spam Message")
```
